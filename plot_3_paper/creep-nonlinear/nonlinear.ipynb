```python
import numpy as np
import matplotlib.pyplot as plt

files = ["8pi.out", "16pi.out", "32pi.out", 
         "64pi.out", "128pi.out", "256pi.out", "512pi.out", "1024pi.out"]

fig, ax = plt.subplots(figsize=(8, 5))

# Plot original files
for f in files:
    data = np.loadtxt(f)
    ax.plot(data[:, 0], data[:, 1], lw=2, label=f)

# Load 8pi once
data8 = np.loadtxt("8pi.out")
t = data8[:, 0]
u = data8[:, 1]

# Multiples of 8pi
c = [1, 2, 4, 8, 16, 32, 64, 128]

for m in c:
    ax.plot(
        t,
        u/m,
        "--",
        lw=1.8,
        label=f"{m}×8pi.out"
    )

ax.set_xlabel("t")
ax.set_ylabel("x(t)/c")
ax.legend()
ax.grid(True, alpha=0.3)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",

    "font.size": 30,
    "axes.labelsize": 30,
    "xtick.labelsize": 30,
    "ytick.labelsize": 30,
    "legend.fontsize": 30,
    "lines.linewidth": 2.0,
    "axes.linewidth": 1.0,
})


files = [
    "8pi.out",
    "16pi.out",
    "32pi.out",
    "64pi.out",
    "128pi.out",
    "256pi.out",
    "512pi.out",
    "1024pi.out"
]

c_vals = [1, 2, 4, 8, 16, 32, 64, 128]

fig, ax = plt.subplots(figsize=(8, 6))

for f, c in zip(files, c_vals):
    data = np.loadtxt(f)

    t = data[:, 0]
    x = data[:, 1]

    ax.plot(
        t,
        x / c,
        label=rf"$c={c}$"
    )

ax.set_xlabel(r"$t$")
ax.set_ylabel(r"$x(t)/c$")
ax.set_xlim(0, 0.5)
ax.set_xticks([0, 0.25, 0.5])
ax.set_ylim(0, 0.3)
ax.set_yticks([0, 0.1, 0.2, 0.3])
ax.grid(True, alpha=0.3)
ax.legend(ncol=2)
fig.savefig("creep_nonlinear.pdf", dpi=300, bbox_inches="tight")

plt.show()
```
